**menu():** Provide a continuous menu for the user.

```python
def menu():
  print("\n==== Grade Management System ====")
  print("1. Add Student Grade")
  print("2. Show all Grades")
  print("3. Show Specific Student's Grade")
  print("4. Update Grade")
  print("5. Delete Grade")
  print("6. Exit")
  while True:
    try:
      option = int(input("Choose an option (1-6):"))
      break
    except ValueError:
      print("Please enter integer type menu option.")

  return option
```

**add_grade():** Add a new student and their grade to the dictionary.

```python
def add_grade(name_grade):
  name = input("Please enter your name:")

  while True:
    try:
      grade = int(input("Please enter your grade:"))
      break
    except ValueError:
      print("Please enter integer type grades.")

  name_grade.update({name: grade})
  print(f"{name}'s grade added successfully!")

  return name_grade
```

**show_all_grades():** Display all students and their grades.

```python
def show_all_grades(name_grade_dict):
  if name_grade_dict is None:
    print("No grades available.")
  else:
    print("--- All Grades ---")
    for item in name_grade_dict:
      print(item,": ", name_grade_dict[item])
```

**show_specific_grade():** Show the grade for a specific student.

```python
def show_specific_grade(name_grade_dict):
  s_name = 1
  student_name = input("Enter the student's name:")
  for item in name_grade_dict:
    if student_name == item:
      s_name=1 #if name is found later in the dictionary then the count should reset
      break
    else:
      s_name = s_name+1 #count of student names not found

  if s_name == 1:
    print(student_name, "'s grade: ", name_grade_dict[student_name])
  else:
    print("Student not found.")
```

**update_grade():** Update the grade of an existing student.

```python
def update_grade(name_grade_dict):
  update_name = 1
  student_name = input("Enter the student's name to update:")
  for item in name_grade_dict:
    if student_name == item:
      update_name=1
      break
    else:
      update_name = update_name+1 #count of student names not found

  if update_name == 1:
    update_grade = int(input("Enter the new grade: "))
    name_grade_dict[student_name] = update_grade
    print(f"{student_name}'s grade updated successfully!")
  else:
    print("Student not found.")
```

**delete_grade():** Delete a student's grade record.

```python
def delete_grade():
  del_name = 1
  student_name = input("Enter the student's name to delete:")
  for item in name_grade_dict:
    if student_name == item:
      del_name=1
      break
    else:
      del_name = del_name+1

  if del_name == 1:
    del name_grade_dict[student_name]
    print(f"{student_name}'s record deleted successfully!")
  else:
    print("Student not found.")
```

```python
option = 1
name_grade = {}

while(option >=1 and option<=6):
  option = menu()
  if option == 1:
    name_grade_dict = add_grade(name_grade)
  elif option == 2:
    show_all_grades(name_grade_dict)
  elif option == 3:
    show_specific_grade(name_grade_dict)
  elif option == 4:
    update_grade(name_grade_dict)
  elif option == 5:
    delete_grade()
  elif option == 6:
    print("Exiting program. Goodbye!")
    break
  else:
    print("Invalid Option!", end="\n\n")
    option = menu()
```
